```python
#建立4個學生的list 
students = [
    [87, 76, 65, 78, 92],  # student1
    [88, 77, 66, 79, 93],  # student2
    [89, 78, 67, 80, 94],  # student3
    [90, 79, 68, 81, 95]   # student4
]
#使用for迴圈取出每個學生的成績
for student in students:
    print(student)
    print("數學成績:", student[2])  # 假設數學成績在索引2的位置
    print("歷史成績:", student[3])  # 假設歷史成績在索引3的位置
    print("===========")  # 分隔每個學生的成績輸出
#這樣就可以一次性處理多個學生的成績了


```
